```python
import gspread
from oauth2client.service_account import ServiceAccountCredentials
import pandas as pd

# Define the scope
scope = ['https://spreadsheets.google.com/feeds', 'https://www.googleapis.com/auth/drive']

# Authenticate using the credentials JSON
creds = ServiceAccountCredentials.from_json_keyfile_name('/Users/mike/Documents/GitHub/No-Kill-Colorado-Data/google_secret.json', scope)
client = gspread.authorize(creds)

# Google Sheet URL
spreadsheet_url = 'https://docs.google.com/spreadsheets/d/1zlh79aLdicsDVjZoxfwVm8SY1iFlKm-GBI34TMk1I30/edit'

# Open the Google Sheet
spreadsheet = client.open_by_url(spreadsheet_url)

# Get all worksheets
worksheets = spreadsheet.worksheets()

# Initialize a list to hold dataframes
dfs = []

for worksheet in worksheets:
    # Get all records as list of dictionaries
    data = worksheet.get_all_records()
    # Convert to pandas DataFrame
    df = pd.DataFrame(data)
    dfs.append(df)

# Now `dfs` is a list of pandas DataFrames, each representing a sheet
# For example, print the first DataFrame
print(dfs[0])
```

```python
import pandas as pd

data_dict = {'Name': ['Alice Ann', 'Bob Smith', 'Charlie Jo Anderson'], 'Age': [25, 30, 35]}
data_df = pd.DataFrame(data_dict)
data_df['First_Name'] = data_df['Name'].str.split().str[0]
data_df['Last_Name'] = data_df['Name'].str.split().str[-1]
syn_dict = {'Name': ['Charlie Jo Anderson'], 'First_Name': ['Charlie Jo'], 'Last_Name': ['Anderson']}
syn_df = pd.DataFrame(syn_dict)

merged_df = pd.merge(data_df, syn_df, on=['Name'], how='left', indicator=True)
merged_df
```
